```python
import pandas as pd
from bs4 import BeautifulSoup
import requests

url = "https://www.icc-cricket.com/rankings/team-rankings/mens/odi"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

page = requests.get(url, headers=headers)
soup = BeautifulSoup(page.content, "html.parser")

# 1. Keep storage lists completely clean and pluralized
positions = []
teams = []
matches = []
points = []
ratings = []

# 2. Iterate explicitly through rows, not the master table shell container
ranking_rows = soup.find_all("div", class_="si-table-row")

for row in ranking_rows:
    # --- POSITION ---
    pos_container = row.find("div", class_="si-table-data si-pos")
    if pos_container is not None and pos_container.find("span", class_="si-text"):
        pos_text = pos_container.find("span", class_="si-text").text.strip()
        positions.append(pos_text)
    else:
        positions.append(None)

    # --- TEAM ---
    team_container = row.find("div", class_="si-table-data si-team")
    if team_container is not None and team_container.find("h3", class_="si-team-name"):
        team_text = team_container.find("h3", class_="si-team-name").text.strip()
        teams.append(team_text)
    else:
        teams.append(None)

    # --- MATCHES ---
    match_container = row.find("div", class_="si-table-data si-matches")
    if match_container is not None and match_container.find("span", class_="si-text"):
        match_text = match_container.find("span", class_="si-text").text.strip()
        matches.append(match_text)
    else:
        matches.append(None)

    # --- POINTS ---
    points_container = row.find("div", class_="si-table-data si-points")
    if points_container is not None and points_container.find("span", class_="si-text"):
        points_text = points_container.find("span", class_="si-text").text.strip()
        points.append(points_text)
    else:
        points.append(None)

    # --- RATINGS ---
    rating_container = row.find("div", class_="si-table-data si-rating")
    if rating_container is not None and rating_container.find("span", class_="si-text"):
        rating_text = rating_container.find("span", class_="si-text").text.strip()
        ratings.append(rating_text)
    else:
        ratings.append(None)

# 3. Create the data frame out of your clean plural lists
ranking_data = pd.DataFrame(
    {
        "Position": positions,
        "Team": teams,
        "Matches": matches,
        "Points": points,
        "Ratings": ratings,
    }
)

# 4. Clean up completely empty rows if any headers were picked up
ranking_data = ranking_data.dropna(how="all").reset_index(drop=True)

print(ranking_data)
```

```python
import pandas as pd
from bs4 import BeautifulSoup

# Simulating your page content loading step
# page = requests.get(url, headers=headers)
# soup = BeautifulSoup(page.content, 'html.parser')

# Pluralized lists to keep your storage completely clean
positions = []
team_shorts = []
team_fulls = []
matches_list = []
points_list = []
ratings_list = []

# 1. Target the body container to avoid picking up table header rows
table_body = soup.find("div", class_="si-table-body")

if table_body:
    # 2. Extract individual data rows safely
    ranking_rows = table_body.find_all("div", class_="si-table-row")

    for row in ranking_rows:
        # --- POSITION ---
        pos_el = row.find("div", class_="si-pos")
        pos_txt = (
            pos_el.find("span", class_="si-text").text.strip() if pos_el else "N/A"
        )
        positions.append(pos_txt)

        # --- TEAM ---
        team_el = row.find("h3", class_="si-team-name")
        if team_el:
            # Captures 'IND' or 'NZ'
            short_txt = (
                team_el.find("span", class_="si-sname").text.strip()
                if team_el.find("span", class_="si-sname")
                else "N/A"
            )
            # Captures 'India' or 'New Zealand'
            full_txt = (
                team_el.find("span", class_="si-fname").text.strip()
                if team_el.find("span", class_="si-fname")
                else "N/A"
            )
        else:
            short_txt, full_txt = "N/A", "N/A"

        team_shorts.append(short_txt)
        team_fulls.append(full_txt)

        # --- MATCHES ---
        match_el = row.find("div", class_="si-matches")
        match_txt = (
            match_el.find("span", class_="si-text").text.strip()
            if match_el
            else "N/A"
        )
        matches_list.append(match_txt)

        # --- POINTS (Fixed from si-points to si-pts) ---
        points_el = row.find("div", class_="si-pts")
        points_txt = (
            points_el.find("span", class_="si-text").text.strip()
            if points_el
            else "N/A"
        )
        points_list.append(points_txt)

        # --- RATINGS ---
        rating_el = row.find("div", class_="si-rating")
        rating_txt = (
            rating_el.find("span", class_="si-text").text.strip()
            if rating_el
            else "N/A"
        )
        ratings_list.append(rating_txt)

# 3. Create the Pandas Dataframe using the extracted values
ranking_data = pd.DataFrame(
    {
        "Position": positions,
        "Abbreviation": team_shorts,
        "Team Name": team_fulls,
        "Matches": matches_list,
        "Points": points_list,
        "Rating": ratings_list,
    }
)

print(ranking_data)
```
