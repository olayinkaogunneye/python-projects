### Imports & Settings

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import os
os.makedirs("../reports/visuals", exist_ok=True)


pd.set_option("display.max_columns", None)
sns.set(style="whitegrid", palette="viridis")

GOLD_DIR = "../data/gold"
```

### Load Gold Tables

```python
dim_book = pd.read_csv(f"{GOLD_DIR}/dim_book.csv")
dim_category = pd.read_csv(f"{GOLD_DIR}/dim_category.csv")
dim_rating = pd.read_csv(f"{GOLD_DIR}/dim_rating.csv")
dim_date = pd.read_csv(f"{GOLD_DIR}/dim_date.csv")
fact = pd.read_csv(f"{GOLD_DIR}/fact_book_metrics.csv")

dim_book.head(), dim_category.head(), dim_rating.head(), dim_date.head(), fact.head()
```

### Validate Row Counts

```python
print("dim_book rows:", len(dim_book))
print("dim_category rows:", len(dim_category))
print("dim_rating rows:", len(dim_rating))
print("dim_date rows:", len(dim_date))
print("fact rows:", len(fact))
```

### Validate Keys & Uniqueness

```python
print("Unique book_id in fact:", fact["book_id"].nunique())
print("Unique category_id in fact:", fact["category_id"].nunique())
print("Unique rating_id in fact:", fact["rating_id"].nunique())
print("Unique date_id in fact:", fact["date_id"].nunique())
```

### Join Fact + Dimensions (Analysis DataFrame)

```python
df = (
    fact
    .merge(dim_book, on="book_id", how="left")
    .merge(dim_category, on="category_id", how="left")
    .merge(dim_rating, on="rating_id", how="left")
    .merge(dim_date, on="date_id", how="left")
)

df.head()
```

### Basic Data Overview

```python
df.info()
df.describe(include="all")
```

### Price Distribution Plot

```python
plt.figure(figsize=(10,5))
sns.histplot(df["price"], bins=20, kde=True)
plt.title("Price Distribution of Books")
plt.xlabel("Price (£)")
plt.ylabel("Count")
plt.show()

plt.savefig("../reports/visuals/price_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Rating Distribution Plot

```python
plt.figure(figsize=(8,4))
sns.countplot(x="rating_value", data=df)
plt.title("Rating Distribution")
plt.xlabel("Rating (1–5)")
plt.ylabel("Count")
plt.show()

plt.savefig("../reports/visuals/rating_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Category Counts

```python
plt.figure(figsize=(12,6))
df["category_name"].value_counts().sort_values().plot(kind="barh")
plt.title("Books per Category")
plt.xlabel("Count")
plt.ylabel("Category")
plt.show()

plt.savefig("../reports/visuals/books_per_category.png", dpi=300, bbox_inches="tight")
plt.show()
```
