## Category‑Level Analysis

### Analysis DataFrame

```python
df = (
    fact
    .merge(dim_book, on="book_id", how="left")
    .merge(dim_category, on="category_id", how="left")
    .merge(dim_rating, on="rating_id", how="left")
    .merge(dim_date, on="date_id", how="left")
)

df.head()
```

### Category Summary Table

```python
category_summary = (
    df.groupby("category_name")
      .agg(
          total_books=("book_id", "count"),
          avg_price=("price", "mean"),
          min_price=("price", "min"),
          max_price=("price", "max"),
          avg_rating=("rating_value", "mean"),
          total_stock=("availability", "sum")
      )
      .sort_values("total_books", ascending=False)
)

category_summary
```

### Number of Books per Category

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=category_summary.reset_index(),
    x="total_books",
    y="category_name",
    palette="viridis"
)
plt.title("Number of Books per Category")
plt.xlabel("Total Books")
plt.ylabel("Category")
plt.show()
```

### Average Price per Category

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=category_summary.reset_index(),
    x="avg_price",
    y="category_name",
    palette="magma"
)
plt.title("Average Price per Category")
plt.xlabel("Average Price (£)")
plt.ylabel("Category")
plt.show()
```

### Average Rating per Category

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=category_summary.reset_index(),
    x="avg_rating",
    y="category_name",
    palette="coolwarm"
)
plt.title("Average Rating per Category")
plt.xlabel("Average Rating (1–5)")
plt.ylabel("Category")
plt.show()
```

### Total Stock per Category

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=category_summary.reset_index(),
    x="total_stock",
    y="category_name",
    palette="crest"
)
plt.title("Total Availability per Category")
plt.xlabel("Total Stock")
plt.ylabel("Category")
plt.show()
```

### Identify Low‑Stock Categories

```python
low_stock = (
    category_summary[category_summary["total_stock"] < 20]
    .sort_values("total_stock")
)

low_stock
```
