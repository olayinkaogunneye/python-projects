```python
# 1.  Load Libraries & Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os 

# Load dataset
netflix_df = pd.read_csv('netflix_data.csv')

# 2. Inspect the Dataset

netflix_df.shape
netflix_df.head(10)
netflix_df.columns
netflix_df.dtypes

# 3. Clean Column Names

def clean_text (text, lower = True):
    text = text.replace( " ", "_")
    return text.lower() if lower else text

netflix_df.columns = [clean_text(col) for col in netflix_df.columns]

# 4. Clean & Convert Data Types

# Convert date_added to datetime

netflix_df['date_added'] = pd.to_datetime(netflix_df['date_added'], errors = 'coerce')

# convert release_year to integer

netflix_df['release_year'] = netflix_df['release_year'].astype(int)

# convert duration to numeric

netflix_df['duration'] = pd.to_numeric(netflix_df['duration'], errors = 'coerce')

# 5. Handle Missing Values

netflix_df['country'] = netflix_df['country'].fillna("Unknown")
netflix_df['director'] = netflix_df['director'].fillna("Unknown")
netflix_df['cast'] = netflix_df['cast'].fillna("Unknown")
netflix_df = netflix_df.dropna(subset=['duration'])

# 6.Create Useful Subsets

# movies only 

movies = netflix_df[netflix_df['type'] == 'Movie']

# 1900s movies

movies_1990s = movies[(movies['release_year'] >= 1990) & (movies['release_year'] <= 1999)]

# Add decade column

movies['decade'] = (movies['release_year'] // 10) * 10

# 7. Quick Data Quality Checks

movies['genre'].unique()
movies['country'].unique()[:20]
netflix_df.duplicated().sum()
netflix_df.isna().sum()
```
